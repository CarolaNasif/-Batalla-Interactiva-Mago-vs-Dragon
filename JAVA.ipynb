```python
%%writefile Batalla.java
class Personaje {
    String nombre;
    int vida;
    int ataque;

    Personaje(String nombre, int vida, int ataque) {
        this.nombre = nombre;
        this.vida = vida;
        this.ataque = ataque;
    }

    void atacar(Personaje enemigo) {
        enemigo.vida -= this.ataque;
        if (enemigo.vida < 0) enemigo.vida = 0; // evitar vida negativa
        System.out.println(nombre + " ataca a " + enemigo.nombre + " causando " + ataque + " de daño ⚔️");
    }

    void mostrarVida() {
        System.out.println(nombre + " tiene " + vida + " puntos de vida ❤️");
    }
}

class Mago extends Personaje {
    Mago(String nombre) {
        super(nombre, 100, 25);
    }

    void lanzarhechizo(Personaje enemigo) {
        System.out.println(nombre + " lanza un hechizo mágico 💥");
        atacar(enemigo);
    }
}

class Dragon extends Personaje {
    Dragon(String nombre) {
        super(nombre, 150, 30);
    }
}

public class Batalla {
    public static void main(String[] args) {
        Mago mago = new Mago("Julio");
        Dragon dragon = new Dragon("Viserion");

        System.out.println("🔥 ¡Comienza la batalla entre " + mago.nombre + " y " + dragon.nombre + "! 🔥\n");

        // Ciclo de batalla
        while (dragon.vida > 0 && mago.vida > 0) {
            mago.lanzarhechizo(dragon);
            dragon.mostrarVida();

            if (dragon.vida <= 0) break; // El dragón muere

            dragon.atacar(mago);
            mago.mostrarVida();

        }

        System.out.println("\n🏆 ¡" + mago.nombre + " ha ganado la batalla! El dragón " + dragon.nombre + " ha sido derrotado. 🐉💀");
    }
}
```

```python
# batalla.py

def crear_personaje(nombre, vida, ataque):
    return {"nombre": nombre, "vida": vida, "ataque": ataque}


def atacar(atacante, enemigo):
    enemigo["vida"] -= atacante["ataque"]
    if enemigo["vida"] < 0:
        enemigo["vida"] = 0
    print(f"{atacante['nombre']} ataca a {enemigo['nombre']} causando {atacante['ataque']} de daño ⚔️")


def mostrar_vida(personaje):
    print(f"{personaje['nombre']} tiene {personaje['vida']} puntos de vida ❤️")


def lanzar_hechizo(mago, enemigo):
    print(f"{mago['nombre']} lanza un hechizo mágico 💥")
    atacar(mago, enemigo)


def main():

    mago = crear_personaje("Julio", 100, 25)
    dragon = crear_personaje("Viserion", 150, 30)

    print(f"🔥 ¡Comienza la batalla entre {mago['nombre']} y {dragon['nombre']}! 🔥\n")


    while dragon["vida"] > 0 and mago["vida"] > 0:
        lanzar_hechizo(mago, dragon)
        mostrar_vida(dragon)

        if dragon["vida"] <= 0:
            break

        atacar(dragon, mago)
        mostrar_vida(mago)

    print(f"\n🏆 ¡{mago['nombre']} ha ganado la batalla! "
          f"El dragón {dragon['nombre']} ha sido derrotado. 🐉💀")


if __name__ == "__main__":
    main()
```
